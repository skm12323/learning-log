### ex 1

```python
for i in range(1,31):
    if i % 15 == 0: # 若为15倍数直接短路
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

### ex 2

```python
odd = []
even = []
nums = [12, 5, 8, 131, 44, 3, 27, 15]
for i in nums:
    if i % 2 == 0:
        even += [i]
    if i % 2 == 1 and i >= 10:
        odd += [i]

even1 = [i for i in nums if i % 2 == 0]
odd1 = [i for i in nums if i % 2 == 1 and i >= 10]

print(even, odd, even1, odd1)
```

### ex 3

```python
text = "quantum mechanics says quantum states evolve under quantum rules"
freq = {}
for word in text.split():
    freq[word] = freq.get(word, 0) + 1

print(max(freq, key=freq.get), freq.get(max(freq, key=freq.get)))

top3 = sorted(freq.items(), key=lambda kv: -kv[1])[:3] 
print(top3)
```

### ex 8

```python
num = int(input("输入整数（0 结束）: "))
nums = []
while(num != 0):
    nums += [num]
    num = int(input("输入整数（0 结束）: "))
print(len(nums) - 1, sum(nums), max(nums))
```

```python
num = int(input("输入整数（0 结束）: "))
max = num
sum = 0
n = 0
while(num != 0):
    n += 1
    sum += num
    if num > max:
        max = num
    num = int(input("输入整数（0 结束）: "))
print(n,sum,max)
```
